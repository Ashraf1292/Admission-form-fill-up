```python
!pip install -q shap xgboost scikit-learn pandas numpy matplotlib

import os
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, roc_auc_score,
    confusion_matrix, roc_curve, ConfusionMatrixDisplay
)

from xgboost import XGBClassifier
import shap

warnings.filterwarnings("ignore")

RANDOM_SEED = 42
np.random.seed(RANDOM_SEED)

FIG_DIR = "figures"
os.makedirs(FIG_DIR, exist_ok=True)

print("Environment ready. Random seed fixed at", RANDOM_SEED)



CSV_PATH = "instagram_accounts.csv"  # <-- EDIT THIS to your actual file path/name

df_raw = pd.read_csv(CSV_PATH)
print(f"Loaded '{CSV_PATH}' with shape: {df_raw.shape}")



print("=" * 60)
print("DATA TYPES")
print("=" * 60)
print(df_raw.dtypes)

print("\n" + "=" * 60)
print("MISSING VALUES PER COLUMN")
print("=" * 60)
missing = df_raw.isna().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No missing values found.")

print("\n" + "=" * 60)
print("DUPLICATE ROWS")
print("=" * 60)
n_dupes = df_raw.duplicated().sum()
print(f"{n_dupes} exact duplicate rows found.")
if n_dupes > 0:
    df_raw = df_raw.drop_duplicates().reset_index(drop=True)
    print(f"Dropped duplicates. New shape: {df_raw.shape}")

print("\n" + "=" * 60)
print("TARGET CLASS DISTRIBUTION (raw labels)")
print("=" * 60)
print(df_raw["class"].value_counts())
print(df_raw["class"].value_counts(normalize=True).round(4))

# %%
# Visualization 1: class distribution
fig, ax = plt.subplots(figsize=(5, 4), dpi=200)
counts = df_raw["class"].value_counts()
ax.bar(counts.index.map({"f": "Fake (f)", "r": "Authentic (r)"}), counts.values,
       color=["#d9534f", "#5cb85c"])
for i, v in enumerate(counts.values):
    ax.text(i, v + max(counts.values) * 0.01, str(v), ha="center", fontweight="bold")
ax.set_ylabel("Number of accounts")
ax.set_title("Class Distribution: Fake vs Authentic Accounts")
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/01_class_distribution.png", dpi=300, bbox_inches="tight")
plt.show()


class_counts = df_raw["class"].value_counts()
imbalance_ratio = class_counts.max() / class_counts.min()
print(f"Imbalance ratio (majority:minority) = {imbalance_ratio:.2f} : 1")

if imbalance_ratio > 3:
    print("=> Notable imbalance detected. We will use class_weight='balanced' (LR, RF) and "
          "scale_pos_weight (XGBoost), computed from the TRAINING split only, rather than SMOTE. "
          "This directly counters imbalance in the loss function without fabricating synthetic rows.")
    USE_CLASS_WEIGHTING = True
else:
    print("=> Imbalance is mild. No resampling or class weighting will be applied — baseline "
          "hyperparameters are used as-is.")
    USE_CLASS_WEIGHTING = False


df = df_raw.copy()

# Binary target: 1 = fake, 0 = authentic
df["target"] = df["class"].map({"f": 1, "r": 0})
assert df["target"].isna().sum() == 0, "Unexpected class label found outside {'f','r'} — inspect df['class'].unique()."

FEATURE_COLS = ["pos", "flw", "flg", "bl", "pic", "lin", "cl", "cz", "ni",
                 "erl", "erc", "lt", "hc", "pr", "fo", "cs", "pi"]
FEATURE_COLS = [c for c in FEATURE_COLS if c in df.columns]
missing_expected = set(["pos", "flw", "flg", "bl", "pic", "lin", "cl", "cz", "ni",
                         "erl", "erc", "lt", "hc", "pr", "fo", "cs", "pi"]) - set(FEATURE_COLS)
if missing_expected:
    print(f"WARNING: expected feature columns not found in file and excluded: {missing_expected}")

X_full = df[FEATURE_COLS].copy()
y_full = df["target"].copy()

print(f"Feature matrix shape: {X_full.shape}")
print(f"Features used: {FEATURE_COLS}")

# %%
# Identify binary/categorical vs numerical features empirically (by observed unique values),
# rather than assuming from the variable name alone.
candidate_binary = []
candidate_numeric = []
for col in FEATURE_COLS:
    n_unique = X_full[col].nunique(dropna=True)
    unique_vals = set(X_full[col].dropna().unique())
    if n_unique <= 2 and unique_vals.issubset({0, 1, 0.0, 1.0, True, False}):
        candidate_binary.append(col)
    else:
        candidate_numeric.append(col)

print("Detected binary/categorical features:", candidate_binary)
print("Detected numerical features:", candidate_numeric)

# %%
# Invalid / extreme value inspection (reported, not auto-removed).
print("Descriptive statistics for numerical features:")
display_stats = X_full[candidate_numeric].describe().T
print(display_stats)

# Flag (but do not drop) negative values in count-like fields, which would be invalid.
count_like = [c for c in ["flw", "flg", "pos"] if c in X_full.columns]
for col in count_like:
    n_negative = (X_full[col] < 0).sum()
    if n_negative > 0:
        print(f"NOTE: {col} has {n_negative} negative values — invalid for a count; "
              f"inspect manually. Not auto-removed per methodology.")

# Flag extreme values (outside 1st/99th percentile) purely for visibility.
for col in candidate_numeric:
    p1, p99 = X_full[col].quantile([0.01, 0.99])
    n_extreme = ((X_full[col] < p1) | (X_full[col] > p99)).sum()
    if n_extreme > 0:
        pass  # Counts are computed for transparency; no automatic removal is performed.

print("\nExtreme-value scan complete. No rows were removed automatically — "
      "outlier handling would require explicit methodological justification per the study design.")

# %%
# Missing-value handling (only relevant if Section 2 found any).
n_missing_total = X_full.isna().sum().sum()
print(f"Total missing feature values: {n_missing_total}")
if n_missing_total > 0:
    print("Missing values will be median-imputed (numerical) / most-frequent-imputed (binary) "
          "INSIDE the pipeline, fit on the training data only.")
else:
    print("No missing values present — no imputer is strictly required, but one is still "
          "included in the pipeline as a safeguard for robustness/reproducibility.")


X_train, X_test, y_train, y_test = train_test_split(
    X_full, y_full, test_size=0.20, stratify=y_full, random_state=RANDOM_SEED
)

print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")
print("Train class distribution:\n", y_train.value_counts(normalize=True).round(4))
print("Test class distribution:\n", y_test.value_counts(normalize=True).round(4))


from sklearn.impute import SimpleImputer

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
binary_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numeric_transformer, candidate_numeric),
    ("bin", binary_transformer, candidate_binary),
])

# Feature name order after the ColumnTransformer, used later for SHAP labeling.
PROCESSED_FEATURE_NAMES = candidate_numeric + candidate_binary


if USE_CLASS_WEIGHTING:
    n_neg = (y_train == 0).sum()
    n_pos = (y_train == 1).sum()
    scale_pos_weight = n_neg / n_pos
    lr_class_weight = "balanced"
    rf_class_weight = "balanced"
    print(f"scale_pos_weight for XGBoost (train-set derived) = {scale_pos_weight:.3f}")
else:
    scale_pos_weight = 1.0
    lr_class_weight = None
    rf_class_weight = None

models = {
    "Logistic Regression": Pipeline(steps=[
        ("preprocess", preprocessor),
        ("clf", LogisticRegression(max_iter=1000, random_state=RANDOM_SEED,
                                    class_weight=lr_class_weight)),
    ]),
    "Random Forest": Pipeline(steps=[
        ("preprocess", preprocessor),
        ("clf", RandomForestClassifier(n_estimators=200, max_depth=None, random_state=RANDOM_SEED,
                                        class_weight=rf_class_weight, n_jobs=-1)),
    ]),
    "XGBoost": Pipeline(steps=[
        ("preprocess", preprocessor),
        ("clf", XGBClassifier(n_estimators=200, max_depth=6, learning_rate=0.1,
                               scale_pos_weight=scale_pos_weight, random_state=RANDOM_SEED,
                               eval_metric="logloss", use_label_encoder=False)),
    ]),
}


cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_SEED)
print("5-fold stratified CV F1-score (training set only):")
for name, pipe in models.items():
    scores = cross_val_score(pipe, X_train, y_train, cv=cv, scoring="f1")
    print(f"  {name}: mean F1 = {scores.mean():.4f} (+/- {scores.std():.4f})")


results = {}
fitted_models = {}

for name, pipe in models.items():
    pipe.fit(X_train, y_train)
    y_pred = pipe.predict(X_test)
    y_prob = pipe.predict_proba(X_test)[:, 1]

    results[name] = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-score": f1_score(y_test, y_pred),
        "ROC-AUC": roc_auc_score(y_test, y_prob),
    }
    fitted_models[name] = pipe
    print(f"\n{name} — test-set performance:")
    for metric, value in results[name].items():
        print(f"  {metric}: {value:.4f}")

# %%
results_df = pd.DataFrame(results).T.round(4)
print("\n" + "=" * 60)
print("MODEL COMPARISON SUMMARY (test set)")
print("=" * 60)
print(results_df)

# %%
# Visualization 2: confusion matrix for each model
fig, axes = plt.subplots(1, 3, figsize=(15, 4.5), dpi=200)
for ax, (name, pipe) in zip(axes, fitted_models.items()):
    y_pred = pipe.predict(X_test)
    cm = confusion_matrix(y_test, y_pred)
    disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Authentic", "Fake"])
    disp.plot(ax=ax, cmap="Blues", colorbar=False)
    ax.set_title(name)
plt.suptitle("Confusion Matrices by Model")
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/02_confusion_matrices.png", dpi=300, bbox_inches="tight")
plt.show()

# %%
# Visualization 3: ROC curves, all three models overlaid
fig, ax = plt.subplots(figsize=(6, 5), dpi=200)
for name, pipe in fitted_models.items():
    y_prob = pipe.predict_proba(X_test)[:, 1]
    fpr, tpr, _ = roc_curve(y_test, y_prob)
    auc = results[name]["ROC-AUC"]
    ax.plot(fpr, tpr, label=f"{name} (AUC = {auc:.3f})")
ax.plot([0, 1], [0, 1], linestyle="--", color="gray", label="Chance")
ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.set_title("ROC Curves: Model Comparison")
ax.legend(loc="lower right")
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/03_roc_curves.png", dpi=300, bbox_inches="tight")
plt.show()


best_model_name = results_df["F1-score"].idxmax()
print("=" * 60)
print("MODEL SELECTION")
print("=" * 60)
print(results_df.sort_values("F1-score", ascending=False))
print(f"\nSelected model for XAI analysis: **{best_model_name}** "
      f"(F1-score = {results_df.loc[best_model_name, 'F1-score']:.4f}, "
      f"ROC-AUC = {results_df.loc[best_model_name, 'ROC-AUC']:.4f})")

best_pipeline = fitted_models[best_model_name]


preprocess_step = best_pipeline.named_steps["preprocess"]
clf_step = best_pipeline.named_steps["clf"]

X_train_proc = preprocess_step.transform(X_train)
X_test_proc = preprocess_step.transform(X_test)

# Handle sparse output defensively (ColumnTransformer can return sparse arrays).
if hasattr(X_train_proc, "toarray"):
    X_train_proc = X_train_proc.toarray()
if hasattr(X_test_proc, "toarray"):
    X_test_proc = X_test_proc.toarray()

X_train_proc_df = pd.DataFrame(X_train_proc, columns=PROCESSED_FEATURE_NAMES)
X_test_proc_df = pd.DataFrame(X_test_proc, columns=PROCESSED_FEATURE_NAMES)

if best_model_name in ("Random Forest", "XGBoost"):
    explainer = shap.TreeExplainer(clf_step)
    shap_values = explainer.shap_values(X_test_proc_df)
    # Some SHAP/sklearn versions return a list [class0, class1] for binary RF; normalize to class-1 array.
    if isinstance(shap_values, list):
        shap_values = shap_values[1]
elif best_model_name == "Logistic Regression":
    explainer = shap.LinearExplainer(clf_step, X_train_proc_df)
    shap_values = explainer.shap_values(X_test_proc_df)
else:
    raise ValueError(f"No SHAP explainer configured for model type: {best_model_name}")

print(f"SHAP values computed for {best_model_name} on {X_test_proc_df.shape[0]} test examples, "
      f"{X_test_proc_df.shape[1]} features.")


mean_abs_shap = np.abs(shap_values).mean(axis=0)
shap_importance_df = pd.DataFrame({
    "feature": PROCESSED_FEATURE_NAMES,
    "mean_abs_shap": mean_abs_shap
}).sort_values("mean_abs_shap", ascending=False).reset_index(drop=True)

print("=" * 60)
print("RANKED GLOBAL FEATURE IMPORTANCE (mean |SHAP value|)")
print("=" * 60)
print(shap_importance_df.to_string(index=False))

# %%
# Visualization 4: SHAP global feature-importance bar plot
fig = plt.figure(figsize=(7, 6), dpi=200)
shap.summary_plot(shap_values, X_test_proc_df, plot_type="bar", show=False)
plt.title(f"SHAP Global Feature Importance — {best_model_name}")
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/04_shap_global_importance.png", dpi=300, bbox_inches="tight")
plt.show()

# %%
# Visualization 5: SHAP summary/beeswarm plot (magnitude + direction)
fig = plt.figure(figsize=(8, 6), dpi=200)
shap.summary_plot(shap_values, X_test_proc_df, show=False)
plt.title(f"SHAP Summary (Beeswarm) — {best_model_name}")
plt.tight_layout()
plt.savefig(f"{FIG_DIR}/05_shap_beeswarm.png", dpi=300, bbox_inches="tight")
plt.show()


y_test_pred = best_pipeline.predict(X_test)
y_test_reset = y_test.reset_index(drop=True)
y_pred_series = pd.Series(y_test_pred)

correct_fake_idx = y_test_reset[(y_test_reset == 1) & (y_pred_series == 1)].index
correct_auth_idx = y_test_reset[(y_test_reset == 0) & (y_pred_series == 0)].index
misclassified_idx = y_test_reset[y_test_reset != y_pred_series].index

selected_cases = {}
if len(correct_fake_idx) > 0:
    selected_cases["Correctly classified — Fake"] = correct_fake_idx[0]
if len(correct_auth_idx) > 0:
    selected_cases["Correctly classified — Authentic"] = correct_auth_idx[0]
if len(misclassified_idx) > 0:
    selected_cases["Misclassified example"] = misclassified_idx[0]
else:
    print("No misclassified examples found in the test set — model achieved perfect "
          "classification on this split; only correct cases are shown below.")

print("Selected local explanation cases (test-set row indices):")
for label, idx in selected_cases.items():
    print(f"  {label}: index {idx}, true={y_test_reset.loc[idx]}, "
          f"predicted={y_pred_series.loc[idx]}")

# %%
# Visualization 6: SHAP local explanations (waterfall) for each selected case
n_cases = len(selected_cases)
fig, axes = plt.subplots(n_cases, 1, figsize=(9, 4.5 * n_cases), dpi=200)
if n_cases == 1:
    axes = [axes]

expected_value = explainer.expected_value
if isinstance(expected_value, (list, np.ndarray)):
    expected_value = expected_value[1] if len(np.atleast_1d(expected_value)) > 1 else float(np.atleast_1d(expected_value)[0])

for ax, (label, idx) in zip(axes, selected_cases.items()):
    plt.sca(ax)
    exp = shap.Explanation(
        values=shap_values[idx],
        base_values=expected_value,
        data=X_test_proc_df.iloc[idx].values,
        feature_names=PROCESSED_FEATURE_NAMES,
    )
    shap.plots.waterfall(exp, show=False, max_display=10)
    ax.set_title(f"{label} (true={y_test_reset.loc[idx]}, predicted={y_pred_series.loc[idx]})",
                 fontsize=10)

plt.tight_layout()
plt.savefig(f"{FIG_DIR}/06_shap_local_explanations.png", dpi=300, bbox_inches="tight")
plt.show()

# %%
# Print the top contributing features per selected case as plain text, for direct use in a paper.
for label, idx in selected_cases.items():
    print(f"\n--- {label} (row {idx}) ---")
    case_shap = pd.DataFrame({
        "feature": PROCESSED_FEATURE_NAMES,
        "value": X_test_proc_df.iloc[idx].values,
        "shap_value": shap_values[idx],
    }).sort_values("shap_value", key=np.abs, ascending=False)
    print(case_shap.head(6).to_string(index=False))


# ## Section 9 — Reproducibility summary
#
# Consolidated printout of everything needed to reproduce/report the experiment.

# %%
print("=" * 70)
print("REPRODUCIBILITY SUMMARY")
print("=" * 70)
print(f"Random seed: {RANDOM_SEED}")
print(f"Dataset dimensions (after de-duplication): {df.shape}")
print(f"Class distribution (raw): \n{df_raw['class'].value_counts().to_dict()}")
print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")
print(f"Class imbalance ratio: {imbalance_ratio:.2f} : 1  |  "
      f"Class weighting used: {USE_CLASS_WEIGHTING}")
print("\nModel performance (test set):")
print(results_df.sort_values("F1-score", ascending=False))
print(f"\nSelected best model: {best_model_name}")
print("\nRanked SHAP global feature importance:")
print(shap_importance_df.to_string(index=False))
print(f"\nFigures saved to: ./{FIG_DIR}/")
for f in sorted(os.listdir(FIG_DIR)):
    print(f"  - {FIG_DIR}/{f}")
```
