```python
# ============================================================
# FixMatch Semi-Supervised Learning — Full Pipeline
# ============================================================

# ── STEP 0: Install dependencies (run in notebook cell) ─────

!pip install torch torchvision tqdm pandas matplotlib scikit-learn

# ── STEP 1: Imports ─────────────────────────────────────────
from google.colab import drive
drive.mount('/content/drive')
import os
import random

import torch
import torch.nn as nn
import torch.nn.functional as F
import torchvision.models as models
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import (
    ConcatDataset, DataLoader, Subset
)

import matplotlib.pyplot as plt
import pandas as pd
from tqdm import tqdm

# ── STEP 2: Device ──────────────────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# ── STEP 3: Transforms ──────────────────────────────────────
weak_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
])

strong_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ColorJitter(0.4, 0.4, 0.4, 0.2),
    transforms.RandomGrayscale(p=0.2),
    transforms.ToTensor(),
])

# ── STEP 4: Load datasets ───────────────────────────────────
base_path = "/content/drive/MyDrive/100 mV"

segments = [
    '1s_segment_raw',   '1s_segment_trans',
    '2s_segment_raw',   '2s_segment_trans',
    '4s_segment_raw',   '4s_segment_trans',
]

raw_datasets = []

for seg in segments:
    path = os.path.join(base_path, seg, "train_val_photos")
    ds = ImageFolder(root=path, transform=weak_transform)
    raw_datasets.append(ds)
    print(f"{seg} → classes: {ds.classes}, samples: {len(ds)}")

full_dataset = ConcatDataset(raw_datasets)
print("\nTOTAL IMAGES:", len(full_dataset))

# ── STEP 5: Labeled / Unlabeled split ───────────────────────
indices = list(range(len(full_dataset)))
random.seed(42)
random.shuffle(indices)

labeled_ratio = 0.1                          # 10 % labeled
split          = int(len(indices) * labeled_ratio)

labeled_idx   = indices[:split]
unlabeled_idx = indices[split:]

print("Labeled samples  :", len(labeled_idx))
print("Unlabeled samples:", len(unlabeled_idx))

labeled_dataset   = Subset(full_dataset, labeled_idx)
unlabeled_dataset = Subset(full_dataset, unlabeled_idx)

# ── STEP 6: FixMatchUnlabeled wrapper ───────────────────────
class FixMatchUnlabeled(torch.utils.data.Dataset):
    """
    Returns (weak_view, strong_view) for each unlabeled image.

    The underlying dataset already applies weak_transform through
    ImageFolder, so we only need to re-read the raw PIL image and
    apply both transforms ourselves.  We therefore bypass the
    Subset/ConcatDataset transforms by going back to the original
    ImageFolder samples.
    """
    def __init__(self, subset, weak_tf, strong_tf):
        self.subset    = subset
        self.weak_tf   = weak_tf
        self.strong_tf = strong_tf

        # Build a flat list of (file_path, class_idx) across all
        # constituent ImageFolders so we can load raw PIL images.
        self._samples = []
        for ds_idx, global_idx in enumerate(subset.indices):
            # Walk ConcatDataset to find which ImageFolder owns this index
            cumulative = 0
            for folder_ds in subset.dataset.datasets:
                if global_idx < cumulative + len(folder_ds):
                    local_idx = global_idx - cumulative
                    self._samples.append(
                        (folder_ds.samples[local_idx][0],   # file path
                         folder_ds.samples[local_idx][1])   # label (ignored)
                    )
                    break
                cumulative += len(folder_ds)

    def __len__(self):
        return len(self._samples)

    def __getitem__(self, idx):
        from PIL import Image
        path, _ = self._samples[idx]
        img = Image.open(path).convert("RGB")
        return self.weak_tf(img), self.strong_tf(img)


unlabeled_fm = FixMatchUnlabeled(
    unlabeled_dataset, weak_transform, strong_transform
)
print("FixMatchUnlabeled dataset ready ✔  len =", len(unlabeled_fm))

# ── STEP 7: DataLoaders ─────────────────────────────────────
batch_size = 32

labeled_loader = DataLoader(
    labeled_dataset,
    batch_size=batch_size,
    shuffle=True,
    drop_last=True,
    num_workers=2,
    pin_memory=True,
)

unlabeled_loader = DataLoader(
    unlabeled_fm,
    batch_size=batch_size,
    shuffle=True,
    drop_last=True,
    num_workers=2,
    pin_memory=True,
)

# Sanity checks
x, y = next(iter(labeled_loader))
print("Labeled batch   — x:", x.shape, "| y:", y.shape)

x_w, x_s = next(iter(unlabeled_loader))
print("Unlabeled batch — weak:", x_w.shape, "| strong:", x_s.shape)

# ── STEP 8: Model factory ───────────────────────────────────
NUM_CLASSES = 3   # Empty, Single, Double

def build_resnet18():
    m = models.resnet18(weights=None)
    m.fc = nn.Linear(m.fc.in_features, NUM_CLASSES)
    return m.to(device)

# ── STEP 9: Supervised baseline ─────────────────────────────
print("\n===== TRAINING SUPERVISED BASELINE =====\n")

model_sup      = build_resnet18()
optimizer_sup  = torch.optim.Adam(model_sup.parameters(), lr=0.001)
criterion      = nn.CrossEntropyLoss()

sup_loss_history = []
sup_acc_history  = []

EPOCHS = 5

for epoch in range(EPOCHS):
    model_sup.train()
    correct, total, epoch_loss = 0, 0, 0.0

    loop = tqdm(labeled_loader, desc=f"[Supervised] Epoch {epoch+1}/{EPOCHS}")
    for x, y in loop:
        x, y = x.to(device), y.to(device)

        logits = model_sup(x)
        loss   = criterion(logits, y)

        optimizer_sup.zero_grad()
        loss.backward()
        optimizer_sup.step()

        epoch_loss += loss.item()
        _, pred = torch.max(logits, 1)
        total   += y.size(0)
        correct += (pred == y).sum().item()
        loop.set_postfix(loss=f"{loss.item():.4f}")

    acc = 100 * correct / total
    sup_loss_history.append(epoch_loss / len(labeled_loader))
    sup_acc_history.append(acc)
    print(f"  → Loss: {sup_loss_history[-1]:.4f} | Acc: {acc:.2f}%")

# ── STEP 10: FixMatch training ───────────────────────────────
print("\n===== TRAINING FIXMATCH =====\n")

model_fix     = build_resnet18()
optimizer_fix = torch.optim.Adam(model_fix.parameters(), lr=0.001)
threshold     = 0.95
lambda_u      = 1.0

fix_loss_history = []
fix_acc_history  = []

for epoch in range(EPOCHS):
    model_fix.train()
    total_loss = 0.0

    labeled_iter   = iter(labeled_loader)
    unlabeled_iter = iter(unlabeled_loader)
    num_iter       = min(len(labeled_loader), len(unlabeled_loader))

    loop = tqdm(range(num_iter), desc=f"[FixMatch]    Epoch {epoch+1}/{EPOCHS}")
    for _ in loop:
        # Labeled batch
        x_lb, y_lb = next(labeled_iter)
        x_lb, y_lb = x_lb.to(device), y_lb.to(device)

        # Unlabeled batch — weak + strong views
        x_w, x_s = next(unlabeled_iter)
        x_w, x_s = x_w.to(device), x_s.to(device)

        # ── Supervised loss ──────────────────────────
        logits_lb = model_fix(x_lb)
        loss_x    = criterion(logits_lb, y_lb)

        # ── Pseudo labels from weak view ─────────────
        with torch.no_grad():
            probs           = torch.softmax(model_fix(x_w), dim=1)
            max_probs, pseudo_labels = torch.max(probs, dim=1)

        mask = (max_probs >= threshold).float()

        # ── Unsupervised loss on strong view ─────────
        logits_s = model_fix(x_s)
        loss_u   = (
            F.cross_entropy(logits_s, pseudo_labels, reduction='none') * mask
        ).mean()

        # ── Combined loss ─────────────────────────────
        loss = loss_x + lambda_u * loss_u

        optimizer_fix.zero_grad()
        loss.backward()
        optimizer_fix.step()

        total_loss += loss.item()
        loop.set_postfix(loss=f"{loss.item():.4f}")

    fix_loss_history.append(total_loss / num_iter)

    # Evaluate on labeled set (same split as supervised for fair comparison)
    model_fix.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for x, y in labeled_loader:
            x, y = x.to(device), y.to(device)
            _, pred = torch.max(model_fix(x), 1)
            total   += y.size(0)
            correct += (pred == y).sum().item()

    acc = 100 * correct / total
    fix_acc_history.append(acc)
    print(f"  → Loss: {fix_loss_history[-1]:.4f} | Acc: {acc:.2f}%")

# ── STEP 11: Results & Plots ─────────────────────────────────
results = pd.DataFrame({
    "Model":              ["Supervised", "FixMatch"],
    "Best Accuracy (%)":  [max(sup_acc_history),  max(fix_acc_history)],
    "Final Loss":         [min(sup_loss_history),  min(fix_loss_history)],
})

print("\n===== FINAL RESULTS =====")
print(results.to_string(index=False))

# Loss comparison
plt.figure()
plt.plot(sup_loss_history, label="Supervised Loss")
plt.plot(fix_loss_history, label="FixMatch Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Loss Comparison")
plt.legend()
plt.savefig("/content/loss_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: /content/loss_comparison.png")

# Accuracy comparison
plt.figure()
plt.plot(sup_acc_history, label="Supervised Accuracy")
plt.plot(fix_acc_history, label="FixMatch Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.title("Accuracy Comparison")
plt.legend()
plt.savefig("/content/accuracy_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: /content/accuracy_comparison.png")

# Bar chart
plt.figure()
plt.bar(["Supervised", "FixMatch"],
        [max(sup_acc_history), max(fix_acc_history)])
plt.ylabel("Accuracy (%)")
plt.title("Final Model Comparison")
plt.savefig("/content/final_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
print("Saved: /content/final_comparison.png")
```
